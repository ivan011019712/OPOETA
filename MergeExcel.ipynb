```python
####################Combine same format excel data without xlwings.
import os
# import xlwings as xw
import pandas as pd
# import numpy as np
# from datetime import date,timedelta,datetime
# import re
# import warnings

df=pd.read_excel('OPOs/tt_a.xlsx',sheet_name='tt',usecols=[0,1])
df['Model']='tt_a'

df=df[['Model']+[col for col in df.columns if col != 'Model']]


df2=pd.read_excel('OPOs/tt_b.xlsx',sheet_name='tt',usecols=[0,1])
df2['Model']='tt_b'
df2=df2[['Model']+[col for col in df2.columns if col != 'Model']]


df=pd.concat([df,df2],axis=0,ignore_index=True)

#df
##############################################################################################################
###########################to New Excel

#df.to_excel('OPOs/tt_all.xlsx', sheet_name='tt_All', index=False)

#############################################################################################################
###########################Direct to SQL
#import pandas as pd
from sqlalchemy import create_engine
import urllib

# 2. 設定連線字串 (根據你的資料庫修改)
# 如果是 SQL Server (使用 Windows 驗證)
conn_str = (
    "Driver={ODBC Driver 17 for SQL Server};"
    "Server=IVAN-MAGNETA;"
    "Database=Test;"
    "Trusted_Connection=yes;"
)
quoted_conn_str = urllib.parse.quote_plus(conn_str)
engine = create_engine(f"mssql+pyodbc:///?odbc_connect={quoted_conn_str}")

# 3. 直接塞入 SQL
# name: 資料表名稱
# if_exists: 'append' (增加), 'replace' (覆蓋), 'fail' (若存在則報錯)
df.to_sql(name='tt_all', con=engine, if_exists='replace', index=False)

print("資料已成功寫入 SQL！")
```
